```python
from sklearn.datasets import load_diabetes
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
x, y = load_diabetes(return_X_y = True)
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 2)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train , y_train)
y_pred = lr.predict(x_test)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
import numpy as np

class GDregressor:
    def __init__(self, epochs=100, LR=0.01):
        self.coef_ = None
        self.intercept_ = None
        self.LR = LR
        self.epochs = epochs

    def fit(self, x_train, y_train):
        # initialise coefs and intercept
        self.intercept_ = 0
        self.coef_ = np.ones(x_train.shape[1])

        for i in range(self.epochs):
            # prediction using current coef_ and intercept_
            y_pred = self.intercept_ + np.dot(x_train, self.coef_)

            # gradients
            intercept_derv = -2 * np.mean(y_train - y_pred)
            coef_derv = -2 * np.dot((y_train - y_pred), x_train) / x_train.shape[0]

            # update step
            self.intercept_ = self.intercept_ - (self.LR * intercept_derv)
            self.coef_ = self.coef_ - (self.LR * coef_derv)

            # debug print every 100 epochs — remove once confirmed working
            if i % 100 == 0:
                print(f"epoch {i}: intercept={self.intercept_:.4f}, derv={intercept_derv:.4f}")

        return self.intercept_, self.coef_

    def predict(self, x_test):
        return self.intercept_ + np.dot(x_test, self.coef_)
```

```python
# Restart kernel first, then run cells in order

gd = GDregressor(epochs=1000, LR=0.01)
gd.fit(x_train, y_train)
```
