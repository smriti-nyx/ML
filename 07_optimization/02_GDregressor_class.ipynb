```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
```

```python
x, y = make_regression(n_samples = 100 , n_informative = 1 , n_features = 1 , n_targets = 1 , noise = 20 )
```

```python
plt.scatter(x, y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x, y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
m = lr.coef_
#aim is to create a class that predicts the accurate b for a constant m 
```

```python
class GDregressor:
     def __init__(self , LR , epochs):
        self.m = 80.63
        self.b = -120
        self.LR = LR
        self.epochs = epochs 
     def fit(self ,x, y):
        #calculate the b using GD 
        for i in range(self.epochs):
            loss_slope = -2*np.sum( y - self.m*x.ravel() - self.b)
            step_size = self.LR * loss_slope
            self.b = self.b  - step_size 
        print(self.b)


```

```python
gd = GDregressor(0.0001 , 1000)
```

```python
gd.fit(x,y) #closest to og when I changed my LR from 0.1 to 0.001
```
