```python

import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('../data/titanic_train.csv', usecols = ['Age' , 'Fare' , 'Survived', 'Pclass'])
```

```python
df.head()
```

```python
df.isnull().mean()*100 #age has 19 % misisng values
```

```python
x = df.iloc[:, 1:]
y = df.iloc[:, 0]
```

```python
from sklearn.model_selection import train_test_split 
x_train ,x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 2 )
```

```python
x_train.sample(10) #can see the missing values in age 
```

```python
from sklearn.impute import KNNImputer
```

```python
knn = KNNImputer(weights = 'uniform') #with or wothout n_neighbors , the answer is same . Hence no neighbors

x_train_trf = knn.fit_transform(x_train ,y_train)
x_test_trf = knn.transform(x_test)
```

```python
x_train_trf = pd.DataFrame(x_train_trf , columns = x_train.columns)
```

```python
x_test_trf = pd.DataFrame(x_test_trf , columns = x_test.columns)
```

```python
x_train_trf.sample(10)
```

```python
x_train.sample(10)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
```

```python
lr.fit(x_train_trf , y_train)
y_pred = lr.predict(x_test_trf)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test , y_pred)
```

comparison with simple imputer

```python
from sklearn.impute import SimpleImputer 
si = SimpleImputer()
```

```python
x_train_trf2 = si.fit(x_train_trf , y_train)
```

```python
x_test_trf2 = pd.DataFrame(x_test_trf2 , columns = x_test.columns)
```

```python
y_pred2 = lr.predict(x_test_trf2)
```

```python
accuracy_score(y_test ,  y_pred2)
```
