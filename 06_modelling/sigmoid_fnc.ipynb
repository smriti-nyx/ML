```python
import numpy as np
from sklearn.datasets import make_classification 
```

```python
x, y = make_classification(n_samples = 100 , n_features = 2 , n_informative =1 ,n_redundant = 0 , 
                            n_clusters_per_class=1 , n_classes =2 , class_sep =30 , random_state = 41 , hypercube = False)
```

```python
import matplotlib.pyplot as plt 
plt.figure(figsize = (10 , 6))
plt.scatter(x[:,0] , x[:,1] , c= y , cmap = 'winter' , s = 100)
```

```python
def perceptron(x,y):
    x = np.insert(x, 0 ,1 , axis = 1)
    weights = np.ones(x.shape[1])
    lr = 0.01

    for i in range(1000):
        j = np.random.randint(0, 100)
        y_hat = step(np.dot(weights , x[j]))
        weights = weights + lr*(y[j] - y_hat)*x[j]

    return weights[0] , weights[1:]
```

```python
def step(z):
    return 1 if z>=0 else 0
```

```python
intercept_ , coef_ = perceptron(x, y)
```

```python
print(intercept_)
print(coef_)
```

```python
m = -(coef_[0]/coef_[1])
b = - (intercept_/coef_[1])
```

```python
x_input = np.linspace(-3 , 3 ,100)
y_input = m*x_input +b 
```

```python
plt.figure(figsize = (10 ,6))
plt.plot(x_input , y_input , color = 'red' , linewidth =3)
plt.scatter(x[:, 0], x[:,1], c= y , cmap = 'winter' , s = 100)
plt.ylim(-3 , 2)
```

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(x, y)
```

```python
m = -(lor.coef_[0][0]/lor.coef_[0][1])
b = - (lor.intercept_/lor.coef_[0][1])
```

```python
x_input1 = np.linspace(-3, 3, 100)
y_input1 = m*x_input1 +b 
```

```python
plt.figure(figsize = (10 ,6))
plt.plot(x_input1 , y_input1 , color = 'black' , linewidth =3, label = 'logistic')
plt.plot(x_input , y_input , color = 'red' , linewidth =3 , label = 'step')
plt.scatter(x[:, 0], x[:,1], c= y , cmap = 'winter' , s = 100)
plt.legend()
plt.ylim(-3 , 2)
```

```python
#using sigmoid instead of step 
def perceptron(x,y):
    x = np.insert(x, 0 ,1 , axis = 1)
    weights = np.ones(x.shape[1])
    lr = 0.01

    for i in range(1000):
        j = np.random.randint(0, 100)
        y_hat = sigmoid(np.dot(weights , x[j]))
        weights = weights + lr*(y[j] - y_hat)*x[j]

    return weights[0] , weights[1:]
```

```python
intercept_ , coef_ = perceptron(x, y)
```

```python
def sigmoid(z):
    return 1/ (1+ np.exp(-z))
```

```python
m = -(coef_[0]/coef_[1])
b = - (intercept_/coef_[1])
```

```python
x_input2 = np.linspace(-6, 6, 200)
y_input2 = m*x_input2 +b 
```

```python
plt.figure(figsize = (10 ,6))
plt.plot(x_input2 , y_input2 , color = 'brown' , linewidth =3, label = 'sigmoid')
plt.plot(x_input1 , y_input1 , color = 'black' , linewidth =3, label = 'logistic')
plt.plot(x_input , y_input , color = 'red' , linewidth =3 , label = 'step')
plt.scatter(x[:, 0], x[:,1], c= y , cmap = 'winter' , s = 100)
plt.legend()
plt.ylim(-3 , 2)
```
