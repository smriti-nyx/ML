```python
import numpy as np
#for creating classifcation dataset we , use 
from sklearn.datasets import make_classification
```

```python
x, y = make_classification(n_samples = 100 , n_features = 2 , n_informative = 1 , n_redundant = 0 , n_classes = 2 ,
                             n_clusters_per_class=1 , random_state = 41 ,hypercube = False , class_sep = 30 )
```

```python
y
```

```python
import matplotlib.pyplot as plt 
plt.figure(figsize = (10 , 6))
plt.scatter(x[: , 0]  , x[:,1],c = y, cmap = 'winter' ,s = 100)
```

```python
#obtaining all the coefficients 
def perceptron(x,y):
    x = np.insert(x , 0 , 1, axis = 1)
    #initialising weights 
    weights = np.ones(x.shape[1])

    lr = 0.1

    for i in range(1000):
        j = np.random.randint(0 ,100)
        y_hat = step(np.dot(weights , x[j]))
        weights = weights + lr * ( y[j] - y_hat)*x[j]

    return weights[0] ,weights[1:]
        
```

```python
def step(z):
    return 1 if z>=0 else 0
```

```python
intercept_ , coef_ = perceptron(x, y)
```

```python
print(intercept_)
print(coef_)
```

```python
#drawing the regression line for the above classification 
#converting the ouput line general equation int he form of mx+b 

m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m*x_input +b 
```

```python
plt.figure(figsize = (10,6))
plt.plot(x_input , y_input , color= 'red' , linewidth = 3)
plt.scatter(x[:,0] , x[: , 1], c= y , cmap ='winter', s = 100)
plt.ylim(-3,2)
```

perceptron trick is good but scikit learn logistic reg model is better because the perceptron output line stops the moment classification is done but sklearn logistic reg imrpoves until it gets the most perfect line .

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x, y )
```

```python
m = -(lr.coef_[0][0]/lr.coef_[0][1])
b = - (lr.intercept_/lr.coef_[0][1])
```

```python
x_input1 = np.linspace(-3 ,3 , 100)
y_input1 = m*x_input1 + b 
```

```python
plt.figure(figsize = (10 , 6))
plt.plot(x_input1 , y_input1 ,color = 'black' , linewidth =3 , label = 'scikit learn logistic')
plt.plot(x_input , y_input , color = 'red' , linewidth = 3 , label = 'perceptron trick')
plt.scatter(x[:,0] , x[:,1] , c= y , cmap = 'winter' , s = 100)
plt.legend()
plt.ylim(-3 , 2)
```

from the above graph , it can be concluded that scikit learn logistic reg is better
