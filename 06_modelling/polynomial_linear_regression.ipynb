```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
x = 6 * np.random.rand(200,1) - 3
y = 0.8 * x**2 + 0.9*x + 2 + np.random.randn(200, 1)

#y = 11x^2 + 5x+m7 
```

```python
plt.scatter(x , y , color = 'blue')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import r2_score 
```

```python
x_train , x_test , y_train , y_test  = train_test_split(x , y , test_size = 0.2 , random_state = 2)
```

```python
lr = LinearRegression()
lr.fit(x_train , y_train)
```

```python
y_pred = lr.predict(x_test)
r2_score(y_test, y_pred) #poor r2 score 
```

```python
plt.plot(x_train, lr.predict(x_train), color = 'red')
plt.scatter(x , y , color = 'blue')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
from sklearn.pipeline import Pipeline
```

```python
#creating a fnc which plots the curve on just givin the degree
def polynomial_regression(degree) :

    x_new = np.linspace(-3, 3, 100).reshape(100 , 1)
    x_new_poly = poly.transform(x_new)


    polybig_features = PolynomialFeatures(degree = degree, include_bias = False)
    std_scaler = StandardScaler()
    lr = LinearRegression()
    polynomial_regression = Pipeline([
        ('poly features' , polybig_features),
        ('std_scaler' , std_scaler),
        ('linear_reg', lr)
    ])

    polynomial_regression.fit(x, y)
    y_newbig = polynomial_regression.predict(x_new)
    plt.plot(x_new , y_newbig , 'r' , label = 'Degree' ,linewidth = 2)


   # plt.plot(x_train , y_train , 'b' , linewidth =  3)
    plt.scatter(x_test , y_test , color = 'blue', linewidth = 3 )
    plt.scatter(x_train , y_train , color = 'green' ,linewidth = 3)
    plt.legend(loc = 'upper left')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.show()
```

```python
plt.style.use('dark_background')
```

```python
polynomial_regression(2) #cross val score can help with finding the right degree
```

```python
#3d polynomial regression 
x =  7 *np.random.rand(100 , 1) -2.8
y =  7 *np.random.rand(100 , 1) -2.8

z = x**2 + y**2 + 0.2*x + 0.2*y + 0.1*x*y +2 + np.random.randn(100, 1)
# z = x^2 + y^2 + 0.2x + 0.2y + 0.1xy + 2

```

```python
import plotly.express as px 
df = px.data.iris()
fig = px.scatter_3d(df , x = x.ravel() ,y = y.ravel(), z= z.ravel())
fig.show()
```

```python
lr =  LinearRegression()
lr.fit(np.array([x, y]).reshape(100 , 2), z)

x_input = np.linspace(x.min() , x.max() , 10)
y_input = np.linspace(x.min() , x.max() , 10)

xGrid , yGrid = np.meshgrid(x_input , y_input )

#final = np.vstack(xGrid.ravel().reshape(1 , 100) , yGrid.ravel().reshape(1, 100)).T
final = np.vstack((xGrid.ravel().reshape(1,100),yGrid.ravel().reshape(1,100))).T


z_final = lr.predict(final).reshape(10 , 10 )
```

```python
import plotly.graph_objects as go

fig = px.scatter_3d(df, x=x.ravel(), y=y.ravel(), z=z.ravel())

fig.add_trace(go.Surface(x = x_input, y = y_input, z =z_final ))

fig.show()
```
